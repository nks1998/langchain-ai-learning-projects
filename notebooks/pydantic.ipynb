```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class UserWithType(BaseModel):
    name: str = Field(max_length=7)
    age: int = Field(gt=0, le=110)
```

```python
user1 = UserWithType(name="Nihal", age=46)
print(user1)
```

```python
#annotated
commonValidator = Annotated[List[UserWithType], "This is common Validator"]
```
