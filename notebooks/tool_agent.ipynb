```python
from dotenv import load_dotenv
load_dotenv()
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
@tool
def add(a: int, b:int):
    """
        it will return the sum of two numbers

        Args:
            a: first number
            b: second number
    """
    return a+b

@tool
def multiple(a: int, b:int):
    """
        it will return the multiplication of two numbers

        Args:
            a: first number
            b: second number
    """
    return a*b
```

```python
add.invoke({"a":3,"b":5})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
out = StrOutputParser()
```

```python
agent = create_agent(
    model=llm,
    tools=[add, multiple],
    system_prompt="You are a math teacher, and always use tool for calculation."
)
```

```python
res = agent.invoke({"messages": [{"role":"user", "content":"What is 2 + 3 * 20 ?"}]})
```
