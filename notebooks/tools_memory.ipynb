```python
from dotenv import load_dotenv
load_dotenv()
from langchain.agents import create_agent
from langchain_core.tools import tool
from langgraph.checkpoint.memory import MemorySaver
```

```python
@tool
def add(a: int, b:int):
    """
        it will return the sum of two numbers

        Args:
            a: first number
            b: second number
    """
    return a+b
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
agent = create_agent(
    model=llm,
    tools=[add],
    system_prompt="You are a math teacher, and always use tool for calculation.",
    checkpointer=MemorySaver()
)
```

```python
question = "IPL winner on 2020?"
question2 = "Total Runner on that match?"
res = agent.invoke(
    {"messages":[{"role":"user", "content":question2}]},
    {"configurable":{"thread_id":"1"}},
)
```
