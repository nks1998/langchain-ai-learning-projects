```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
out = StrOutputParser()
```

```python
def transform(result:str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content":"You are a translator and translate input in {language}"},
    {"role":"user", "content": "{query}"}
])
# final_prompt = prompts.format_messages(language="spanish", query="I love Javascript, and Python")
# res = llm.invoke(prompts.invoke({"language":"spanish", "query":"I love Javascript, and Python"}))

chain = prompts | llm | out | transform
res = chain.invoke({"language":"Hingish", "query":"I love Javascript, and Python"})
```
