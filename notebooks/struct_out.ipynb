### Structure Output

```python
from pydantic import BaseModel, Field

text = "Hello, My name is NIhal Sharma,Email is nks@gmail.com and my age is 24"

class ResponseStructure(BaseModel):
    name: str = Field(description="Complete Name")
    email: str = Field(description="Email Address")
    age: int = Field(description="Age")

structure_llm = llm.with_structured_output(ResponseStructure)
res = structure_llm.invoke(f"Find name, age and email fr0m {text}")

print(res.name)
print(res.email)
print(res.model_dump)
```

```python
from typing import List
class Movies(BaseModel):
    title: str = Field(description="Movie Name")
    year: int = Field(description="Releasing Year")

class MovieList(BaseModel):
    movies: List[Movies]

movie_llm = llm.with_structured_output(MovieList)
res = movie_llm.invoke("Give me 4 movie staring Tom chruse")
print(res.movies)
```
